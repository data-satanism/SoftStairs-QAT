```python
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt

from pathlib import Path
```

```python
from pathlib import Path
OUTPUT = Path("outputs")
for run_dir in OUTPUT.iterdir():
    csv = run_dir / "metrics.csv"
    if csv.exists():
        csv.unlink()
        print("deleted", csv)
```

```python
FOLDER = Path("../../experiments/ebm/outputs").resolve()
baseline = f"baseline-no-qat-{20}e"   # тот же N_EPOCHS, что в train.ipynb

def read_data(run):
    df = pd.read_csv(FOLDER / run / "metrics.csv")
    df["epoch"] = pd.to_numeric(df["epoch"], errors="coerce")
    return df
```

```python
from itertools import product
def compare_runs(run, baseline, metrics=None):
    if metrics is None:
        metrics = [
            ("train_loss", "Train loss"),
            ("val_cdiv", "Val CDIV"),
            ("val_precision", "Precision (uniform fake)"),
            ("val_recall", "Recall (uniform fake)"),
            ("val_f1", "F1 (uniform fake)"),
            ("t", "Quantization t"),
        ]
    run_df = read_data(run)
    base_df = read_data(baseline)

    fig, axes = plt.subplots(2, 3, figsize=(14, 8))
    for ax, (col, title) in zip(axes.flatten(), metrics):
        if col not in run_df.columns:
            ax.set_title(f"{title} (missing)")
            continue
        sns.lineplot(run_df, x="epoch", y=col, ax=ax, label=run)
        if col in base_df.columns:
            sns.lineplot(base_df, x="epoch", y=col, ax=ax, label=baseline)
        ax.set_title(title)
        ax.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
compare_runs(
    "linear-0.9-4b-20e-wd0.0",
    baseline,
)
```
